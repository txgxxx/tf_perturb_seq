```python
# Match ChromBPNet/TF-MoDISco motif patterns to enhancer regions.
# This uses the motif report created by chrombpnet.motif.sh:
#   modisco report -i CM_ATAC_modisco_results.h5 -o modisco_report -s modisco_report -m motifs.meme.txt

motif_report_html = DATA_PATH / "modisco_report" / "motifs.html"
motif_qval_cutoff = 0.05

motif_report = pd.read_html(motif_report_html)[0]

def motif_match_to_tf_name(motif_match):
    """Return a readable TF/TF-complex name from a motif database match ID."""
    motif_match = str(motif_match)
    # Examples: CTCF_MA0139.1 -> CTCF; TEAD1_HUMAN.H11MO.0.A -> TEAD1;
    # FOSB+JUN_MA1127.1 -> FOSB+JUN; NKX2-3_homeodomain_2 -> NKX2-3.
    return motif_match.split("_")[0]

motif_annotations = []
for rank in range(3):
    tmp = motif_report[["pattern", f"match{rank}", f"qval{rank}"]].copy()
    tmp = tmp.rename(columns={f"match{rank}": "motif_match", f"qval{rank}": "motif_qval"})
    tmp["match_rank"] = rank
    motif_annotations.append(tmp)

motif_annotations = pd.concat(motif_annotations, ignore_index=True)
motif_annotations = motif_annotations.dropna(subset=["motif_match", "motif_qval"])
motif_annotations["motif_name"] = motif_annotations["pattern"].str.replace("pos_patterns.", "", regex=False)
motif_annotations["tf_name"] = motif_annotations["motif_match"].map(motif_match_to_tf_name)
motif_annotations["tf_symbols"] = motif_annotations["tf_name"].map(lambda tf: tf.split("+"))

significant_motif_annotations = motif_annotations[
    motif_annotations["motif_qval"] <= motif_qval_cutoff
].copy()

best_motif_annotation = (
    significant_motif_annotations
    .sort_values(["motif_name", "motif_qval", "match_rank"])
    .drop_duplicates("motif_name")
)

region_bed_motif_named = region_bed_motif.copy()
region_bed_motif_named.columns = [
    "chrom", "start", "end", "peak_id", "score", "strand",
    "signal_value", "p_value", "q_value", "summit_offset", "motif_name",
]
region_bed_motif_named["enhancer_id"] = (
    region_bed_motif_named["chrom"].astype(str) + ":" +
    region_bed_motif_named["start"].astype(str) + "-" +
    region_bed_motif_named["end"].astype(str)
)

# One row per significant TF motif match per enhancer. If a motif pattern has several
# significant database matches, this keeps them all because they often represent TF families.
tfmotif_enhancer_links = (
    region_bed_motif_named
    .merge(significant_motif_annotations, on="motif_name", how="inner")
    .sort_values(["chrom", "start", "end", "motif_name", "motif_qval", "match_rank"])
    .drop_duplicates([
        "enhancer_id", "peak_id", "motif_name", "tf_name", "motif_match", "match_rank"
    ])
)

# Optional stricter table: only the best TF motif match per ChromBPNet motif pattern.
tfmotif_enhancer_links_best = (
    region_bed_motif_named
    .merge(best_motif_annotation, on="motif_name", how="inner")
    .sort_values(["chrom", "start", "end", "motif_name"])
    .drop_duplicates(["enhancer_id", "peak_id", "motif_name"])
)

tfmotif_enhancer_links.to_csv(
    output_path + "tfmotif_enhancer_links.q0.05.tsv", sep="\t", index=False
)
tfmotif_enhancer_links_best.to_csv(
    output_path + "tfmotif_enhancer_links.best_match.q0.05.tsv", sep="\t", index=False
)

# display(motif_annotations[["motif_name", "motif_match", "tf_name", "motif_qval", "match_rank"]].head(20))
# tfmotif_enhancer_links.head()
```

```python
# Link TF motif-enhancer calls to downstream genes using positional overlap with scE2G elements.
# Result: TF motif -> ChromBPNet enhancer -> overlapping scE2G element -> downstream gene.

sce2g_path = Path(
    "/home2/s440862/.synapseCache/802/173000802/"
    "igvf3_wtc11_cardio_cardiomyocyte_scE2G_multiome_powerlaw_v3_threshold0.177.e2g.tsv.gz"
)
sce2g = pd.read_csv(sce2g_path, sep="\t", comment="#")

chrombpnet_enhancers = (
    tfmotif_enhancer_links[
        ["enhancer_id", "chrom", "start", "end", "peak_id"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

sce2g_elements = (
    sce2g[
        [
            "ElementChr", "ElementStart", "ElementEnd", "ElementName", "ElementClass",
            "GeneSymbol", "GeneEnsemblID", "GeneTSS", "SampleSummaryShort",
            "Score", "Score.ignoreTPM", "isSelfPromoter",
        ]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

def overlap_chrombpnet_to_sce2g(enhancers, elements):
    """Return all half-open BED overlaps between ChromBPNet enhancers and scE2G elements."""
    overlaps = []

    for chrom, enh_chr in enhancers.groupby("chrom", sort=False):
        elem_chr = elements[elements["ElementChr"] == chrom].copy()
        if elem_chr.empty:
            continue

        enh_chr = enh_chr.sort_values(["start", "end"]).reset_index(drop=True)
        elem_chr = elem_chr.sort_values(["ElementStart", "ElementEnd"]).reset_index(drop=True)
        element_starts = elem_chr["ElementStart"].to_numpy()

        for enhancer in enh_chr.itertuples(index=False):
            # BED intervals are half-open: overlap if element_start < enhancer_end
            # and element_end > enhancer_start.
            last_possible = element_starts.searchsorted(enhancer.end, side="left")
            candidates = elem_chr.iloc[:last_possible]
            candidates = candidates[candidates["ElementEnd"] > enhancer.start]

            if candidates.empty:
                continue

            candidates = candidates.copy()
            candidates["enhancer_id"] = enhancer.enhancer_id
            candidates["chrombpnet_chrom"] = enhancer.chrom
            candidates["chrombpnet_start"] = enhancer.start
            candidates["chrombpnet_end"] = enhancer.end
            candidates["peak_id"] = enhancer.peak_id
            candidates["overlap_bp"] = (
                candidates["ElementEnd"].clip(upper=enhancer.end)
                - candidates["ElementStart"].clip(lower=enhancer.start)
            )
            overlaps.append(candidates)

    if not overlaps:
        return pd.DataFrame()

    return pd.concat(overlaps, ignore_index=True)

enhancer_sce2g_overlaps = overlap_chrombpnet_to_sce2g(
    chrombpnet_enhancers, sce2g_elements
)

tfmotif_enhancer_gene_links = (
    tfmotif_enhancer_links
    .merge(
        enhancer_sce2g_overlaps,
        on=["enhancer_id", "peak_id"],
        how="inner",
    )
    .sort_values(
        [
            "tf_name", "GeneSymbol", "Score", "motif_qval",
            "chrom", "start", "end", "match_rank",
        ],
        ascending=[True, True, False, True, True, True, True, True],
    )
    .drop_duplicates(
        [
            "tf_name", "motif_name", "motif_match", "enhancer_id", "peak_id",
            "ElementName", "GeneSymbol", "GeneEnsemblID",
        ]
    )
)

tfmotif_enhancer_gene_links_best = (
    tfmotif_enhancer_links_best
    .merge(
        enhancer_sce2g_overlaps,
        on=["enhancer_id", "peak_id"],
        how="inner",
    )
    .sort_values(
        ["tf_name", "GeneSymbol", "Score", "motif_qval", "chrom", "start", "end"],
        ascending=[True, True, False, True, True, True, True],
    )
    .drop_duplicates(
        ["tf_name", "motif_name", "enhancer_id", "peak_id", "ElementName", "GeneSymbol", "GeneEnsemblID"]
    )
)

tfmotif_enhancer_gene_links.to_csv(
    output_path + "tfmotif_enhancer_gene_links.sce2g_overlap.q0.05.tsv",
    sep="\t",
    index=False,
)
tfmotif_enhancer_gene_links_best.to_csv(
    output_path + "tfmotif_enhancer_gene_links.sce2g_overlap.best_match.q0.05.tsv",
    sep="\t",
    index=False,
)

print(f"ChromBPNet enhancers with TF motifs: {chrombpnet_enhancers.shape[0]:,}")
print(f"scE2G links: {sce2g.shape[0]:,}")
print(f"enhancer-scE2G positional overlaps: {enhancer_sce2g_overlaps.shape[0]:,}")
print(f"TF motif-enhancer-gene links: {tfmotif_enhancer_gene_links.shape[0]:,}")
# display(
#     tfmotif_enhancer_gene_links[
#         [
#             "tf_name", "motif_name", "motif_match", "motif_qval",
#             "enhancer_id", "peak_id", "ElementName", "ElementClass",
#             "overlap_bp", "GeneSymbol", "GeneEnsemblID", "GeneTSS", "Score",
#             "isSelfPromoter",
#         ]
#     ].head(20)
# )
```
